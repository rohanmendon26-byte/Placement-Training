```python
import numpy as np
```

### Aggregrate function


```python
arr=np.array([10,20,30,40,50])
print("sum:",np.sum(arr))
print("max:",np.max(arr))
print("min:",np.min(arr))
print("count:",arr.size)
print("cumullative sum:",np.cumsum(arr))
print("cumullative product:",np.cumprod(arr))
```

### Statistical Function

```python
import statistics as st

sales=np.array([200,180,150,250,270,300,250,230,500,1000])
print("mean:",np.mean(sales))
print("median:",np.median(sales))
print("mode:",st.mode(sales))
print("Standard deviation:",np.std(sales))
```
