### Numpy Array Operations

```python
import numpy as np
```

```python
arr=np.array([10,20,30,40,50])
print("Basic Slicing:",arr[:4])
print("Slicing with step:",arr[0:5:2])
print("Negetive Index:",arr[-1])
```

```python
arr_2d=np.array([[1,2,3,10],
                 [4,5,6,20],
                 [7,8,9,30]])
print("Specific element:",arr_2d[1,2])
# print("Entire row:",arr_2d[1:,1:])  
# print("Entire column",arr_2d[0:,1]) 
print(arr_2d[2,:2])  
```

# Sorting

```python
arr_1d=np.array([3,5,2,9,5,6,1,7])
print("Sorting 1d array",sorted(arr_1d))
print("Sorting 1d array",np.sort(arr_1d))

arr_2d=np.array([[2,1],
                 [3,2],
                 [1,3]])
print("Sorting 2d array\n",np.sort(arr_2d,axis=1)) #sort the row
print("Sorting 2d array\n",np.sort(arr_2d,axis=0)) #sort the column
```

## Filter

```python
number=np.array([3,5,2,9,5,6,1,7])
even_number=number[number%2==0]
print(even_number)
```

## Filter with mask

```python
arr = np.array([10, 20, 30, 40, 50])
mask=arr>45
print(arr[mask])
```

### Index

```python
arr = np.array([10, 20, 30, 40, 50])
idx = [0, 2, 4] 
print(arr[idx]) 
```

```python
import numpy as np
arr1=np.array([1,2,3,4])
arr2=np.array([10,20,30,40])
combined=np.concatenate([arr1,arr2])
print("Combined:",combined)
```

```python
matrix=np.array([[10,20],
                [30,40]])
newMatrix=np.array([[50,60],
                    [70,80]])
print("join Horizontal:\n",np.hstack([matrix,newMatrix]))
print("join vertically:\n",np.vstack([matrix,newMatrix]))
```

### Adding and Removing

```python
arr=np.array([1,2,3,4,5,6])
print("split:",np.split(arr,2))
print("Append:",np.append(arr,50))
print("Insert:",np.insert(arr,0,90))
print("Remove:",np.delete(arr,2))
print("Org:",arr)
```
