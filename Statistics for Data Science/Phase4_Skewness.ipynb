## Measure of Shape

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
student_marks={
    "s_id":["S01","S02","S03","S04","S04","S06","S06","S08","S09","S10","S11","S12","S13","S14"],
    "Marks":[80,80,85,85,85,90,90,90,90,95,95,95,100,100]
}

student_df=pd.DataFrame(student_marks)

student_df['Marks'].skew()
```

```python
student_df['Marks'].mean()
```

```python
sns.histplot(x="Marks",data=student_df,bins=[i for i in range(70,130,5)])
plt.xticks(range(70,130,10))
```

### Positive Skewness(Right Skewed Distribution)

```python
titanic_df=pd.read_csv("titanic.csv")
titanic_df.describe()
```

```python
titanic_df['Fare'].skew()
```

```python
titanic_df['Fare'].mean(),titanic_df['Fare'].median(),titanic_df['Fare'].mode()[0]

```

```python
sns.histplot(x='Fare',data=titanic_df)
```
