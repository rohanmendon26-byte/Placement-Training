```python
#get
get_values=student.get("Age")
print(get_values)

#keys
dict_keys=student.keys()
print(dict_keys)

#values
dict_values=student.values()
print(dict_values)

#update
student.update({"Age":10})
print(student)

#setdefault
student.setdefault("dept","Cse")
student.setdefault("Name","Anonymous")
print(student)

#copy
c=student.copy()
print(c)

#pop - Remove value based on key
student.pop("dept")
print(student)

#items
print(student.items())
```

write a script to generate a lookup dictionary where keys represent risk tiers (1 to 15) and values are the squares of each tier number.

```python
lookup_dictionary={}
for i in range(1,16):
    lookup_dictionary.setdefault(i,i**2)
print(lookup_dictionary)
```
