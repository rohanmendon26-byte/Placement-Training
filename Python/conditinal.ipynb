```python
age=int(input("Enter your age"))
if age<=3:
    print("Child")
elif age>=4 and age<=12:
    print("Kid")
elif age>=13 and age<=18:
    print("Teenage")
elif age>=19 and age<=60:
    print("Adult")
else:
    print("Ajja")
```
