```python
#Type 1 - Using paranthesis
nums=(1,2,3,40)
color=("Red","Blue","Green",)
mixed=("Raj",23,54.8,True)
nested=(23,(2,4),[2,5,6])

print(nums)
print(color)
print(mixed)
print(nested)
```

```python
#Type 2 - Without Using parenthesis
fruits="Orange","Apple","Banana"
print(type(fruits))

#Type 3 - Using tuple()cons
values=tuple((1,2,3,4,5))
print(values,type(values))
list_item=[10,20,30,40]
list_tup=tuple(list_item)

#Type 4
num=(45,)
print(num,type(num))
```
