### Compare,Pivoting & Melting

```python
import pandas as pd
```

### Compare

```python
shop_7Sep={
    "Fruites":["Mango","Apple","Banana","Papaya"],
    "Prices":[100,150,60,30],
    "Quantity":[10,8,12,14]
}

stock_7Sep=pd.DataFrame(shop_7Sep)

shop_8Sep={
    "Fruites":["Mango","Apple","Banana","Papaya"],
    "Prices":[120,200,60,35],
    "Quantity":[15,6,24,17]
}

stock_8Sep=pd.DataFrame(shop_8Sep)

stock_7Sep.compare(stock_8Sep)
stock_7Sep.compare(stock_8Sep,align_axis=0)
stock_7Sep.compare(stock_8Sep,keep_shape=True)
stock_7Sep.compare(stock_8Sep,keep_equal=True)
stock_7Sep.compare(stock_8Sep,align_axis=0,keep_equal=True,keep_shape=True)
```

### Pivoting

```python
monthly_sales={
    "Product":["Laptop","Mobile","Laptop","Mobile","Laptop","Laptop"],
    "Month":["Jan","Jan","Feb","Feb","Mar","Apr"],
    "Price":[95000,25000,67000,12000,65000,75000]
}

sales=pd.DataFrame(monthly_sales)
sales

#-------Pivoting----------
sales.pivot(
    columns="Month",
    index="Product",
    values="Price"
)
```

```python
monthly_sales={
    "Product":["Laptop","Mobile","Laptop","Mobile","Laptop","Laptop","Mobile"],
    "Month":["Jan","Jan","Feb","Feb","Mar","Mar","Feb"],
    "Price":[95000,25000,67000,12000,65000,75000,10000]
}

sales=pd.DataFrame(monthly_sales)
sales

#-------Pivoting----------
sales.pivot_table(
    columns="Month",
    index="Product",
    values="Price",
    aggfunc="sum"
)
```

### Melting

```python
shop_sales={
    "Product":["Laptop","Mobile","Mouse"],
    "Jan":[45000,14000,140],
    "Feb":[55000,23000,199],
    "Mar":[56000,12000,149]
}

sales_df=pd.DataFrame(shop_sales)
sales_df.melt(
    id_vars="Product",
    var_name="Month",
    value_name="Price"
)  
```
